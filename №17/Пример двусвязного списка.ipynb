```python

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, data):
        """Добавление элемента в конец списка"""
        new_node = Node(data)

        if self.head is None: # Если список пуст
            self.head = new_node
            self.tail = new_node
        else:
            # Связываем новый узел с текущим хвостом
            self.tail.next = new_node
            new_node.prev = self.tail
            # Обновляем хвост
            self.tail = new_node

    def print_forward(self):
        """Вывод от начала к концу"""
        current = self.head
        while current:
            print(current.data, end=" <-> ")
            current = current.next
        print("None")

    def print_backward(self):
        """Вывод от конца к началу (преимущество двусвязного списка)"""
        current = self.tail
        while current:
            print(current.data, end=" <-> ")
            current = current.prev
        print("None")

# Использование
dll = DoublyLinkedList()
dll.append(1)
dll.append(2)
dll.append(3)

dll.print_forward()   # 1 <-> 2 <-> 3 <-> None
dll.print_backward()  # 3 <-> 2 <-> 1 <-> None
```

```python

from collections import deque
# использование collections.deque
# deque реализована как двусвязный
#  список блоков
# примеры использования

d = deque()
d.append(1)       # Добавить в конец
d.appendleft(0)   # Добавить в начало (быстро!)
d.pop()           # Удалить с конца
d.popleft()       # Удалить с начала (быстро!)
```
